```python
import sys
DATA_DIR = '/content/drive/MyDrive/DATASETS/VTE_dataset/participant_split_raw_data'
OUTPUT_DIR = '/content/drive/MyDrive/DATASETS/VTE_dataset/audio_embeddings_data/3sec'
CHECKPOINT_DIR ='/content/drive/MyDrive/DATASETS/VTE_dataset/audio_embeddings_data/checkpoint'

sys.path.append('/content/drive/MyDrive/VTE_classification/feature_extraction')

from dataset_configs import vte_dataset_config
from utils import extract_dataset_features

vte_dataset = vte_dataset_config(
    root_dir=DATA_DIR,
    frame_duration=3.0,
    sample_rate=16000,
    overlap_percentage=0.25
)
```

# VGGish
